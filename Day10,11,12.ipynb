```python
#creating sample data
x= np.linspace(0,10,100) #100 points from 0 to 10  eg print even numbers from 0 to 10 100 times
y= np.sin(x)
 #simple linbe plot
plt.figure(figsize=(8,5))  
plt.plot(x,y)
plt.xlabel('X values')
plt.ylabel('sin(X)')
plt.title('Sine Wave')
plt.grid(True) #add grid for easier reading
plt.show
```

```python
#multile line on the same plot
x =np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)
y3=np.sin(x)*np.cos(x)

plt.figure(figsize=(10,6))
plt.plot(x,y1,label='sin(x)')
plt.plot(x,y2,label='cos(x)')
plt.plot(x,y3,label='sin(x)*cos(x)')
plt.xlabel('X values')
plt.ylabel('Y va;lues')
plt.title('Multiple functions')
plt.legend()  # show legend
plt.grid(True,alpha=0.3)    #alpha control tarnsparency
```

```python
#customizing plots 
#different line size and markers
x =np.linspace(0,10,20)
plt.plot(figsize=(12,6))
plt.plot(x,x,'b-', label='solid line',linewidth=2)
plt.plot(x,x+1,'r--',label='Dashed line',linewidth=2)
plt.plot(x,x+2,'g-',label='dash-Dot line',linewidth=2)
plt.plot(x,x+3,'m',label='Dotted line',linewidth=2)
plt.plot(x ,x+4, 'co-', label='Circles', markersize=8)
plt.plot(x,x+5 ,'s-', label='Squares' , markersize=6)
plt.plot(x,x+6,'^-',label='Triangles',markersize=6)
plt.xlabel('X values')
plt.ylabel('Y va;lues')
plt.title('Line style and markers')
plt.legend()    #to show the box in which lines are deifned
plt.grid(True,alpha=0.3)
plt.show()

```

```python
#customizing with easy commands
x=np.linspace(0,10,100)

plt.figure(figsize=(10,6))
plt.plot(x,np.sin(x),color='steelblue',linewidth=2,label='steelblue')
plt.plot(x,np.cos(x),color='#FF6B6B',linewidth=2,label='#FF6B6B(hex)')
plt.plot(x,np.sin(x)*0.5,color='Darkgreen',linestyle='--',linewidth=2,label='darkgreen')

plt.xlabel('X values',fontsize=12)
plt.ylabel('y value',fontsize=12)
plt.title('custom colors and styles', fontsize=14,fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True,alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(46)      #seed in which our random values stores 
x=np.random.randn(100)  #random values whose mean value is 0 and standard deviation is 1
y=2*x+np.random.randn(100)*0.5   #linear relationship with noise 

plt.Figure(figsize=(8,6))
plt.scatter(x,y,alpha=0.6)    #alpha controls transparency
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('scatter plot')
plt.grid(True,alpha=0.3)
plt.show()
```

```python
np.random.seed(46)      
x=np.random.randn(100)  
y=np.random.randn(100)
colors=np.random.rand(100)    #random colors
sizes=1000*np.random.rand(100)     #(random sizes)the random values are so small so we multiplied it with 1000


plt.Figure(figsize=(10,6))
plt.scatter(x,y,c=colors,s=sizes,alpha=0.6,cmap='plasma')  #cmap is used to fill colors in the plot
plt.colorbar(label="color intensity")    #adds color bar in the side
plt.xlabel('X values')
plt.ylabel('y values')
plt.title("scatter plot with color and size variations")
plt.grid(True,alpha=0.3)
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[23,44,56,78,32]

plt.Figure(figsize=(10,6))
plt.bar(categories,values,color='steelblue',edgecolor='Black',linewidth=1.2)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title("bar garph")
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
plt.Figure(figsize=(10,6))
plt.barh(categories,values,color='coral',edgecolor='Black',linewidth=1.2)
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title("horizontal bar garph")
plt.grid(True, alpha=0.3, axis='x')
plt.show()
```
